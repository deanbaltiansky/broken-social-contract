```python
# %% -------------------------------------------
# Setup
# ----------------------------------------------
import pandas as pd
import numpy as np
from pathlib import Path

from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import RidgeCV, LassoCV, LinearRegression
import statsmodels.api as sm

# ---- Load your data ----
# If you're in Jupyter and the file is next to the notebook, change this path to "df_formodels.csv"
df = pd.read_csv("~/Google Drive/My Drive/american's dream/for github/broken-social-contract/study-2/data/df_formodels.csv")

value_cols = ["democracy","equality","freedom","individualism",
              "justice","happiness","arms","tolerance"]

control_vars = [
    "ideo_con_z","SDO_z","TIPI_agree_z","man","white","hispanic",
    "income_num_z","edu_num_z","age_z",
    "county_medianincome_z","county_gini_z","county_density_z"
]

outcome_vars = ["antiest_z","change_z","trust_deminst_z","trust_natinst_z"]

# Coerce to numeric just in case
for c in value_cols + control_vars + outcome_vars:
    if c in df.columns and not pd.api.types.is_numeric_dtype(df[c]):
        df[c] = pd.to_numeric(df[c], errors="coerce")

# Small utility
def ensure_dir(p):
    Path(p).mkdir(parents=True, exist_ok=True)
```

```python
# %% -------------------------------------------
# Option A already working (LMG / Shapley among values)
# (kept here to save CSVs alongside ridge/lasso)
# ----------------------------------------------
import statsmodels.api as sm

def r2_of(y, X):
    Xc = sm.add_constant(X, has_constant="add")
    return sm.OLS(y, Xc).fit().rsquared

def lmg_among_values(d, y, value_cols, control_vars, n_perm=500, random_state=123):
    rng = np.random.default_rng(random_state)
    yv = d[y].values
    X_ctrl = d[control_vars] if control_vars else pd.DataFrame(index=d.index)
    base_r2 = r2_of(yv, X_ctrl) if control_vars else 0.0

    vals = value_cols[:]
    contrib_sum = dict.fromkeys(vals, 0.0)

    for _ in range(n_perm):
        ordering = vals.copy()
        rng.shuffle(ordering)
        current_cols = control_vars[:]  # start with controls fixed
        prev_r2 = base_r2

        for v in ordering:
            X_curr = d[current_cols + [v]]
            r2_curr = r2_of(yv, X_curr)
            incr = max(0.0, r2_curr - prev_r2)  # guard against tiny negatives
            contrib_sum[v] += incr
            current_cols = current_cols + [v]
            prev_r2 = r2_curr

    contrib_avg = {v: contrib_sum[v] / n_perm for v in vals}
    total = sum(contrib_avg.values())
    shares = {v: (contrib_avg[v] / total if total > 0 else np.nan) for v in vals}

    out = (pd.DataFrame({"predictor": list(shares.keys()),
                         "share_among_values": list(shares.values())})
             .sort_values("share_among_values", ascending=False)
             .reset_index(drop=True))
    out["n"] = len(d)
    return out
```

```python
# %% -------------------------------------------
# Option B — Ridge & Lasso
# Two flavors:
#   (1) Unpenalize controls  → residualize y and the 8 values on controls, then penalize values.
#   (2) Penalize everything  → penalize values + controls together.
# ----------------------------------------------

def residualize_on_controls(d, y, value_cols, control_vars):
    """
    Partial out controls from the outcome and each value predictor.
    Returns y_res (array) and X_val_res (DataFrame with residualized values).
    """
    # If there are no controls, just return raw
    if not control_vars:
        return d[y].values, d[value_cols].copy()

    dd = d[[y] + value_cols + control_vars].dropna().copy()
    # Residualize y
    lr = LinearRegression().fit(dd[control_vars].values, dd[y].values)
    y_res = dd[y].values - lr.predict(dd[control_vars].values)

    # Residualize each value on controls
    Xv_res = {}
    C = dd[control_vars].values
    for v in value_cols:
        lv = LinearRegression().fit(C, dd[v].values)
        Xv_res[v] = dd[v].values - lv.predict(C)

    Xv_res = pd.DataFrame(Xv_res, index=dd.index)
    return y_res, Xv_res, dd.index  # return index to align later

def fit_ridge_lasso_unpen_controls(d, y, value_cols, control_vars,
                                   ridge_alphas=None, lasso_cv=10, random_state=123):
    """
    Controls are held fixed (partialed out). We fit RidgeCV and LassoCV on the residualized
    value predictors only. Returns two DataFrames with coefficients for values.
    """
    if ridge_alphas is None:
        ridge_alphas = np.logspace(-6, 6, 200)

    # Prepare residualized data
    y_res, Xv_res, keep_idx = residualize_on_controls(d, y, value_cols, control_vars)
    # Standardize + fit Ridge
    ridge = make_pipeline(StandardScaler(with_mean=True, with_std=True),
                          RidgeCV(alphas=ridge_alphas, store_cv_values=False))
    ridge.fit(Xv_res.values, y_res)
    ridge_alpha = ridge.named_steps['ridgecv'].alpha_
    ridge_coefs = ridge.named_steps['ridgecv'].coef_

    ridge_df = pd.DataFrame({
        "predictor": value_cols,
        "coef": ridge_coefs
    }).assign(alpha=ridge_alpha, model="ridge_unpen_controls", outcome=y, n=len(keep_idx))

    # Standardize + fit Lasso (CV chooses alpha)
    lasso = make_pipeline(StandardScaler(with_mean=True, with_std=True),
                          LassoCV(cv=lasso_cv, random_state=random_state, n_jobs=None))
    lasso.fit(Xv_res.values, y_res)
    lasso_alpha = lasso.named_steps['lassocv'].alpha_
    lasso_coefs = lasso.named_steps['lassocv'].coef_
    lasso_selected = (lasso_coefs != 0)

    lasso_df = pd.DataFrame({
        "predictor": value_cols,
        "coef": lasso_coefs,
        "selected": lasso_selected
    }).assign(alpha=lasso_alpha, model="lasso_unpen_controls", outcome=y, n=len(keep_idx))

    return ridge_df, lasso_df

def fit_ridge_lasso_penalize_all(d, y, value_cols, control_vars,
                                 ridge_alphas=None, lasso_cv=10, random_state=123):
    """
    Penalize values and controls together. We still report only the value coefficients.
    """
    if ridge_alphas is None:
        ridge_alphas = np.logspace(-6, 6, 200)

    cols = value_cols + control_vars
    dd = d[[y] + cols].dropna().copy()

    # Ridge
    ridge = make_pipeline(StandardScaler(with_mean=True, with_std=True),
                          RidgeCV(alphas=ridge_alphas, store_cv_values=False))
    ridge.fit(dd[cols].values, dd[y].values)
    ridge_alpha = ridge.named_steps['ridgecv'].alpha_
    all_names = cols
    all_coefs = ridge.named_steps['ridgecv'].coef_
    # only keep values
    ridge_vals = pd.DataFrame({
        "predictor": all_names,
        "coef": all_coefs
    })
    ridge_vals = ridge_vals[ridge_vals["predictor"].isin(value_cols)].copy()
    ridge_vals = ridge_vals.assign(alpha=ridge_alpha, model="ridge_penalize_all", outcome=y, n=len(dd))

    # Lasso
    lasso = make_pipeline(StandardScaler(with_mean=True, with_std=True),
                          LassoCV(cv=lasso_cv, random_state=random_state, n_jobs=None))
    lasso.fit(dd[cols].values, dd[y].values)
    lasso_alpha = lasso.named_steps['lassocv'].alpha_
    all_coefs = lasso.named_steps['lassocv'].coef_
    lasso_vals = pd.DataFrame({
        "predictor": all_names,
        "coef": all_coefs
    })
    lasso_vals = lasso_vals[lasso_vals["predictor"].isin(value_cols)].copy()
    lasso_vals = lasso_vals.assign(
        selected=lambda x: x["coef"] != 0,
        alpha=lasso_alpha, model="lasso_penalize_all", outcome=y, n=len(dd)
    )

    return ridge_vals, lasso_vals
```

```python
# %% -------------------------------------------
# Run everything & export CSVs
# ----------------------------------------------
results_dir = "results_py"
ensure_dir(results_dir)

# how many permutations for LMG (you can raise later)
LMG_PERM = 500
LASSO_CV_FOLDS = 10
RIDGE_GRID = np.logspace(-6, 6, 200)

all_lmg = []
all_ridge_unpen = []
all_lasso_unpen = []
all_ridge_all = []
all_lasso_all = []

for y in outcome_vars:
    # Drop NAs only once per outcome
    d = df[[y] + value_cols + control_vars].dropna().copy()
    if d.empty:
        print(f"[WARN] No complete rows for outcome {y}. Skipping.")
        continue

    # ---- LMG / Shapley (values only, controls fixed) ----
    lmg_df = lmg_among_values(d, y, value_cols, control_vars, n_perm=LMG_PERM, random_state=123)
    lmg_df = lmg_df.assign(outcome=y)
    lmg_df.to_csv(f"{results_dir}/optionA_LMG_{y}.csv", index=False)
    all_lmg.append(lmg_df)

    # ---- Ridge/Lasso: controls unpenalized via residualization ----
    ridge_u, lasso_u = fit_ridge_lasso_unpen_controls(
        d, y, value_cols, control_vars,
        ridge_alphas=RIDGE_GRID, lasso_cv=LASSO_CV_FOLDS, random_state=123
    )
    ridge_u.to_csv(f"{results_dir}/optionB_ridge_unpen_{y}.csv", index=False)
    lasso_u.to_csv(f"{results_dir}/optionB_lasso_unpen_{y}.csv", index=False)
    all_ridge_unpen.append(ridge_u); all_lasso_unpen.append(lasso_u)

    # ---- Ridge/Lasso: penalize everything ----
    ridge_all, lasso_all = fit_ridge_lasso_penalize_all(
        d, y, value_cols, control_vars,
        ridge_alphas=RIDGE_GRID, lasso_cv=LASSO_CV_FOLDS, random_state=123
    )
    ridge_all.to_csv(f"{results_dir}/optionB_ridge_all_{y}.csv", index=False)
    lasso_all.to_csv(f"{results_dir}/optionB_lasso_all_{y}.csv", index=False)
    all_ridge_all.append(ridge_all); all_lasso_all.append(lasso_all)

# Also save combined files for convenience
if all_lmg:
    pd.concat(all_lmg, ignore_index=True).to_csv(f"{results_dir}/ALL_optionA_LMG.csv", index=False)
if all_ridge_unpen:
    pd.concat(all_ridge_unpen, ignore_index=True).to_csv(f"{results_dir}/ALL_ridge_unpen.csv", index=False)
if all_lasso_unpen:
    pd.concat(all_lasso_unpen, ignore_index=True).to_csv(f"{results_dir}/ALL_lasso_unpen.csv", index=False)
if all_ridge_all:
    pd.concat(all_ridge_all, ignore_index=True).to_csv(f"{results_dir}/ALL_ridge_all.csv", index=False)
if all_lasso_all:
    pd.concat(all_lasso_all, ignore_index=True).to_csv(f"{results_dir}/ALL_lasso_all.csv", index=False)

print("Done. CSVs written to:", results_dir)
```
